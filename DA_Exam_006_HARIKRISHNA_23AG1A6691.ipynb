```python
import pandas as pd
```

```python
data=pd.read_csv('/content/car_price_performance.csv')
data
```

```python
br=data.groupby('Brand')['Price'].mean().sort_values(ascending=True)
br
```

```python
dic={}
for i in range(len(br)):
    dic[br.index[i]]=i+1
```

```python
l1=[]
for i in range(len(data)):
    l1.append(dic[data['Brand'][i]])
```

```python
data['Encodede_Brand']=l1
data
```

```python
#MIN-MAX SCALLING
l2=[]
for i in data['Horsepower']:
  l2.append((i-min(data['Horsepower']))/(max(data['Horsepower'])-min(data['Horsepower'])))
data['Scaled_HorsePower']=l2
```

```python
l3=[]
for i in data['Fuel Efficiency']:
  l3.append((i-min(data['Fuel Efficiency']))/(max(data['Fuel Efficiency'])-min(data['Fuel Efficiency'])))
data['Scaled_Fuel_Efficiency']=l3
data
```

```python
l4=[]
for i in data['Encodede_Brand']:
  l4.append((i-min(data['Encodede_Brand']))/(max(data['Encodede_Brand'])-min(data['Encodede_Brand'])))
data['Scaled_Brand']=l4
data
```

```python
#pi=hp*0.7+f*0.3/p
l5=[]
for i in range(len(data)):
  pi=(data['Horsepower'][i]*0.7+data['Fuel Efficiency'][i]*0.3)/data['Price'][i]
  l5.append(pi)
data['Performance_Index']=l5
data
```

```python
data.to_csv("final_data_HariKrishna_691.csv")
```
